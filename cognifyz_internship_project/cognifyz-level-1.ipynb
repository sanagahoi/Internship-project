```python
import pandas as pd

df = pd.read_csv('Dataset.csv')
df.head()
```

- there are **9551** rows and **21** columns present in the dataset

```python
df.isna().sum()
```

- There are 9 missing values in the dataset which we can drop

```python
df1 = df.dropna()
```

```python
df1.info()
```

```python
df1['Aggregate rating'].value_counts()
```

```python
def categorize_rating(col):
    if col == 0:
        return "not rated (0.0)"
    elif col<2.5:
        return 'Poor (< 2.5)'
    elif col < 3.5:
        return "Average (2.5 - 3.5)"
    elif col<4.5:
        return "Good (3.5 - 4.5)"
    else:
        return "excellent (4.5 - 5.0)"
    
df1['rating_range'] = df1['Aggregate rating'].apply(categorize_rating)
df1['rating_range'].value_counts()
```

- Significant imbalance exists. Over 22% of restaurants (2,148) have a rating of 0.0 (unrated/zero rating), while ratings between 2.5 and 3.9 represent the majority of rated places.
