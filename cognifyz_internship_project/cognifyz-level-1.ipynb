###### Task 1 - Data Exploration and Preprocessing

- Explore the dataset and identify the number of rows and columns.
- Check for missing values in each column and handle them accordingly.

- Analyze the distribution of the target variable ("Aggregate rating") and identify any class imbalances.

```python
import pandas as pd

df = pd.read_csv('Dataset.csv')
df.head()
```

```python
# Identify the number of rows and columns.
df.shape
```

- there are **9551** rows and **21** columns present in the dataset

```python
# Check for missing values in each column and handle them accordingly.
df.isna().sum()
```

- There are 9 missing values in the dataset which we can drop

```python
# It's negligible number so we can delete those entries
df1 = df.dropna()
```

```python
# Perform data type conversion if necessary.
df1.info()
```

```python
# Convert identifier and categorical codes to string
df1['Restaurant ID'] = df1['Restaurant ID'].astype(str)
df1['Country Code'] = df1['Country Code'].astype(str)
```

```python
df1.info()
```

```python
# Analyze the distribution of the target variable ("Aggregate rating") and identify any class imbalances.
df1['Aggregate rating'].value_counts()
```

```python
def categorize_rating(col):
    if col == 0:
        return "not rated (0.0)"
    elif col<2.5:
        return 'Poor (< 2.5)'
    elif col < 3.5:
        return "Average (2.5 - 3.5)"
    elif col<4.5:
        return "Good (3.5 - 4.5)"
    else:
        return "excellent (4.5 - 5.0)"
```

```python
df1.loc[:, 'rating_range'] = df1['Aggregate rating'].apply(categorize_rating)
df1['rating_range'].value_counts()
```

- Significant imbalance exists. Over 22% of restaurants (2,148) have a rating of 0.0 (unrated/zero rating), while ratings between 2.5 and 3.9 represent the majority of rated places.

##### Task 3 : Geospatial Analysis

- Visualize the locations of restaurants on a map using latitude and longitude information.
- Analyze the distribution of restaurants across different cities or countries.
- Determine if there is any correlation between the restaurant's location and its rating.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Scatter plot of Longitude vs Latitude
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Longitude', y='Latitude', data=df1, hue='City', palette='viridis', s=10, alpha=0.7)
plt.title('Restaurant Geographic Distribution by Latitude and Longitude')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', ncol=2)
plt.show()
```

```python
# Distribution of restaurants across countries (using country code counts)
country_distribution = df1['Country Code'].value_counts()
print("Top Countries by Restaurant Count:")
print(country_distribution.head())

# Distribution across top cities
city_distribution = df1['City'].value_counts()
print("\nTop 10 Cities by Restaurant Count:")
print(city_distribution.head(10))
```

```python
# Correlation between coordinates and ratings
location_rating_corr = df1[['Longitude', 'Latitude', 'Aggregate rating']].corr()
print("Correlation Matrix:")
print(location_rating_corr)

# Optional: Visualize correlation with a heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(location_rating_corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation: Location vs Aggregate Rating')
plt.show()
```
